## 1. My lane as an ML task (type)

Task Type: Binary Classification

Why: We need to predict whether a specific piece of content/URL requires an immediate refresh (1 = Needs Refresh, 0 = Healthy/No Action Needed). Classification allows us to create a clear, actionable threshold to prioritize editorial resources toward content experiencing true decay.

```python
import pandas as pd

# Load dataset to verify classification target concept
# The original relative path doesn't work in Colab, so load from raw GitHub URL.
df_sample = pd.read_csv("https://raw.githubusercontent.com/Amna-Asif1911/Flyrank-ML-Internship/main/data/raw/content_refresh_anonymized.csv")
print("Total rows available for classification:", len(df_sample))
```

## 2. Target or proxy

Target Proxy: needs_refresh (Binary Flag)

Source: Defined Rule / Proxy. Since we do not have direct human labels for "needs refresh," we construct a proxy label based on observed historical outcomes: if an article's organic traffic drops by more than 20% over a 90-day window (traffic_drop_pct > 0.20), it is labeled as 1 (Needs Refresh), otherwise 0.

```python
# Construct proxy target variable
df_sample["needs_refresh"] = (pd.to_numeric(df_sample["trend_pct"], errors='coerce') < -20).astype(int)

# Print target class distribution
print("Target Class Distribution:")
print(df_sample["needs_refresh"].value_counts(normalize=True))
```

## 3. Success metric

Metric: Precision@K (or Precision at chosen classification threshold)Defensible Target: Precision $\ge$ 0.80 for the top-flagged pages.Why: Editorial time is expensive. If the model flags 50 pages for a rewrite (False Positives), human editors waste time reviewing healthy content. We prioritize high precision to ensure at least 80% of flagged URLs genuinely require content updates.

```python
# Mock metric calculation demonstration
from sklearn.metrics import precision_score

# Example check verifying evaluation strategy
y_true = df_sample["needs_refresh"]
# Baseline dummy check
print(
    f"Baseline target prevalence (positive class ratio): {y_true.mean():.2%}"
)
```

## 4. The unit of analysis, as a real dataframe

Unit of Analysis: 1 Row = 1 Content URL

Each row represents a unique content piece / URL along with its historical traffic, backlink health, age, and word count metrics.

```python
import pandas as pd

# Load starter dataset
# The original relative path doesn't work in Colab, so load from raw GitHub URL.
df = pd.read_csv("https://raw.githubusercontent.com/Amna-Asif1911/Flyrank-ML-Internship/main/data/raw/content_refresh_anonymized.csv")

# Create target column using 'trend_pct' which contains traffic change percentages
df["target_needs_refresh"] = (pd.to_numeric(df["trend_pct"], errors='coerce') < -20).astype(int)

# Display unit of analysis: 1 row = 1 URL
print(f"Dataset Shape: {df.shape} (Rows = URLs, Columns = Features + Target)")
df[["content_id", "trend_pct", "target_needs_refresh"]].head()
```

## 5. Why ML beats a fixed rule here

Why ML Beats Fixed Rules:

A simple rule like if traffic_drop > 20% then refresh fails because traffic drops occur for multiple non-content reasons (e.g., seasonality, site migration issues, search volume drops). Machine learning combines multiple non-linear signals—such as backlink decay rate, competitor word count growth, page age, and search intent shift—to separate true content decay from temporary seasonal noise.

```python
# Feature interaction check showing why a single rule is noisy
correlation = df[
    ["trend_pct", "content_age_days"]
].corr()
print("Feature Correlation Matrix:")
print(correlation)
```
